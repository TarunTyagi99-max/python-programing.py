```python
#smartcampus automated pass & utility tariff calculator
print("=" * 65)
print("smartcampus utility & access pass generator")
print("=" * 65)
try:
    category = int(input("select user category (1: student, 2: faculty/staff): "))
    if category not in (1 , 2):
        print("[error]: please select a valid category")
    else:
        sub = input("enter sub category (ug/pg or resistent/visiting/guest)").strip().lower()
        valid_input = True
        cgpa, exp = 0.0 , 0
    
        if category == 1:
            if sub not in ("ug", "pg"):
                print("[error]: student sub-category must be ug or pg .")
                valid_input = False
            else:
                cgpa = float(input("enter student cgpa (0.0 - 10.0): "))
                if not (0.0 <= cgpa <= 10.0):
                    print("[error]: enter cgpa between 0.0 and 10.0")
                    valid_input = False
                base = 500.0 if sub == "ug" else 350.0
        else:
            if sub not in ("resident", "visiting", "guest"):
                print("[error]: faculty sub-category must be resident, visiting, or guest.")
                valid_input = False
            else:
                exp = int(input("enter year of service: "))
                if exp < 0:
                    print("error")
                    valid_input = False
                base = 800.0 if sub == "resident" else 1200.0
        if valid_input:
            disc, disc_label = 0.0,""
            if category == 1:
                if cgpa >= 8.5:
                    disc, disc_label = base * 0.20, "merit discount (20%)"
                elif cgpa >= 7.5:
                    disc, disc_label = base * 0.10, "merit discount (10%)"
            else:
                if exp > 10:
                    disc , disc_label = base * 0.15, "service discount (15%)"
            p_output = int(input("select parking (0: none, 2: two-wheeler, 4: four-wheeler): ")) 
            rates = {0: (0.0, "none"), 2: (200.0, "two-wheeler"), 4:(600.0, "four-wheeler")}
            if p_output not in rates:
                print("[error]: invalid parking option select.")
            else:
                p_fee, p_name = rates[p_output]
                surcharge = 150.0 if (p_output == 4 and category == 1) else 0.0
            
                units = float(input("ebter monthly consuption (in kwh): "))
                if units < 0:
                    print("[error]: electricity units cannot be nagitive")
                else:
                    def calc_power(u):
                        if u <= 100:
                            return u*3, 50
                        if u <= 300:
                            return 300 + (u - 100) * 5, 100
                        if u <= 500:
                            return 1300 + (u - 300) * 7.5, 150
                        return 2800 + (u - 500) * 10, 50
                    p_charge, fix_charge = calc_power(units)
                    elec_total = p_charge + fix_charge
                
                    net_pass = base - disc + p_fee + surcharge 
                    total = net_pass + elec_total
                
                    print("\n" + "_" * 65)
                    print("calculated invoice breakdown")
                    print("_" * 65)
                    print(f"base access pass fee       :  rupee {base :.2f}")
                
                    if disc > 0:
                        print(f"{disc_label: <27}: - rupee {disc:.2f}")
                    if p_fee > 0:
                        print(f"parking fee ({p_name})".ljust(27) + f": rupee {p_fee :.2f}")
                    if surcharge > 0:
                        print(f"student peak surcharge         : rupee {surcharge :.2f}")   
                
                    print(f"net pass & parking total           : {net_pass:.2f}")
                    print("_" * 65)
                    print(f"total monthly payable amount: {total:.2f}")
                    print("=" * 65)
except ValueError:
    print("[error]: invalid input! please enter valid numeric value")
                
              
                 
                
```
